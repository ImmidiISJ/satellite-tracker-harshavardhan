```python
import math
import time
import random
import folium
from IPython.display import display, clear_output

# ==================================================================
# BASE CONFIGURATION (Safe local calculation sandbox - NO API KEY REQUIRED)
# ==================================================================
MY_LAT = 13.0827
MY_LON = 80.2707
MY_CITY = "Chennai"

print("Initializing Global Mesh Engine...")
print("Registering the entire world map layout inside your browser...")
time.sleep(1)

# Seed the physics scatter array so it remains stable but fully scattered
random.seed(42)

# Generate 800 satellites completely scattered across the whole planet
# No lines, no clusters - a true uniform global mesh covering the entire map
global_fleet = []
for i in range(800):
    if i < 650:
        orbit_type = "LEO (Low-Earth Orbit)"
        alt = 550
        speed_factor = 15.2
    else:
        orbit_type = "MEO (Global Navigation)"
        alt = 20200
        speed_factor = 2.0

    global_fleet.append({
        "id": i,
        "name": f"Global-Node-{orbit_type[:3]}-{i:03d}",
        "mean_motion": speed_factor,
        "altitude_km": alt,
        "start_lat_phase": random.uniform(0, 360),
        "start_lon_phase": random.uniform(0, 360),
        "inclination_angle": random.uniform(15, 85)
    })

try:
    for frame in range(10):
        current_time = time.time()
        readable_time = time.strftime('%Y-%m-%d %H:%M:%S', time.localtime(current_time))

        # FIXED: Switched tiles to OpenStreetMap to remove the API watermark permanently
        m = folium.Map(location=[0, 0], zoom_start=1.5, tiles="OpenStreetMap")

        # Anchor your ground base station marker pin (⌂) on the world map
        folium.Marker(
            location=[MY_LAT, MY_LON],
            popup=f"Your Base Monitoring Station ({MY_CITY})",
            icon=folium.Icon(color="green", icon="home")
        ).add_to(m)

        # Compute physics positions across the full global map coordinates
        for sat in global_fleet:
            period = (24 * 3600) / sat["mean_motion"]

            lat_angle = (2 * math.pi * (current_time % period) / period) + math.radians(sat["start_lat_phase"])
            lon_angle = (2 * math.pi * (current_time % period) / period) + math.radians(sat["start_lon_phase"])

            sat_lat = math.sin(lat_angle) * sat["inclination_angle"]
            sat_lon = math.cos(lon_angle) * 180.0

            sat_lat = max(-85.0, min(85.0, sat_lat))
            sat_lon = ((sat_lon + 180) % 360) - 180

            # Line of Sight Distance calculation to your desk
            rlat1, rlon1 = math.radians(MY_LAT), math.radians(MY_LON)
            rlat2, rlon2 = math.radians(sat_lat), math.radians(sat_lon)
            dlat, dlon = rlat2 - rlat1, rlon2 - rlon1

            a = math.sin(dlat/2)*2 + math.cos(rlat1) * math.cos(rlat2) * math.sin(dlon/2)*2
            a = max(0.0, min(1.0, a))
            c = 2 * math.atan2(math.sqrt(a), math.sqrt(1-a))

            ground_dist = 6371 * c
            direct_dist = math.sqrt(ground_dist*2 + sat["altitude_km"]*2)

            node_color = "#38bdf8" if sat["altitude_km"] < 1000 else "#f59e0b"

            # Render the coordinate dot onto the global world map layer
            folium.CircleMarker(
                location=[sat_lat, sat_lon],
                radius=2.0,
                color=node_color,
                fill=True,
                fill_color=node_color,
                popup=f"<b>{sat['name']}</b><br>Altitude: {sat['altitude_km']} km<br>Distance to Desk: {direct_dist:.2f} km"
            ).add_to(m)

        clear_output(wait=True)
        print(f"🛰️ WHOLE WORLD THEORETICAL RADAR STREAM | TIME: {readable_time} UTC")
        print(f"Master Console Status: Tracking {len(global_fleet)} active nodes spread across the global grid.")
        print("Blue Nodes = Low Earth Orbit | Yellow Nodes = Navigation Networks | Green Pin = Your Monitor Location")
        display(m)

        time.sleep(5)

except KeyboardInterrupt:
    print("\nGlobal radar layout paused successfully.")
```
